# 2. Merge single-Higgs VEV-zero batches

This stage combines the ten JSON outputs of the batch VEV-selection calculation into `killed_vevs_library_single_higgs_models.json`. See the [README](README.md) for how the generation, batch script, and fitting stages connect.

**Inputs:** `killed_vevs_library0.json` through `killed_vevs_library9.json`, in the notebook working directory. They are not included in this repository. Each maps a model identifier to a list of minimal sets of zero-based singlet indices.

## Imports

```python
import json
from pathlib import Path
N_BATCHES = 10
BATCH_DIR = Path('.')
```

## Merge and export

Run after all ten batches have completed. Batch model identifiers should be disjoint; the legacy merge overwrites earlier values if a key occurs in multiple files. The output replaces any existing merged file.

```python
killed_vevs_library = {}
for job in range(N_BATCHES):
    with (BATCH_DIR / f'killed_vevs_library{job}.json').open() as file:
        batch = json.load(file)
    killed_vevs_library.update(batch)
with open('killed_vevs_library_single_higgs_models.json', 'w') as file:
    json.dump(killed_vevs_library, file, indent=2)
```
